When one route fails, the failed invariant names the obstruction.

## 1. Hamiltonian Problem Signature

### Definition 1.1 (Legal Hamiltonian Signature)
A Hamiltonian problem signature is
$$
\mathfrak H=(\mathcal H,H,\mathfrak q_H,\Sigma,k_\Sigma,V_\Sigma,q,E_q,\mathcal B,\mathcal G),
$$
where:

- $\mathcal H$ is a separable Hilbert space. In finite computation, $\mathcal H=L^2(\Omega_n,\mu_n)$.
- $H\ge0$ is self-adjoint with closed quadratic form $\mathfrak q_H$ and form domain $D(\mathfrak q_H)$. When a non-normal pseudospectral row is invoked, the signature also contains a represented closed densely defined operator $A$ on the same reduced Hilbert space, a represented common core when needed, and a declared resolvent window $\Omega_A\subset\rho(A)$.
- $\Sigma$ is the target or accepting set, represented by a target gate $k_\Sigma$ and, when killing is used, a nonnegative killing potential or absorbing form $V_\Sigma$.
- $q$ is a represented quotient and $E_q$ is its lumpability or terminal-sector defect.
- $\mathcal B$ is the budget structure from the task-space framework.
- $\mathcal G$ is the Noether/gauge subspace of conserved directions that do not change the target verdict.
- When noncommutative observable or expectation-gauge rows are invoked, the signature also contains a represented von Neumann algebra $(\mathcal M,\varphi)$ with faithful normal state, an abelian affordable target-readout subalgebra $\mathcal C_\Sigma$, a gauge/readout subalgebra $\mathcal N$ containing the declared target readouts, and a represented $\varphi$-preserving conditional expectation $E_{\mathcal N}:\mathcal M\to\mathcal N$ when such an expectation is not automatic from finite or tracial data.

The killed Hamiltonian is the nonnegative self-adjoint operator
$$
K_\Sigma=H+V_\Sigma
$$
when $V_\Sigma$ is form-bounded with relative bound $<1$, and otherwise the Friedrichs operator associated with the closed killed form
$$
\mathfrak q_{K_\Sigma}(u,u)=\mathfrak q_H(u,u)+\int V_\Sigma |u|^2.
$$
Dirichlet killing on $\Sigma$ is the special case obtained by restricting the form domain to functions whose quasi-continuous representative vanishes on $\Sigma$.

The signature is legal for a requested Hamiltonian fast-track row when the following row hypotheses are supplied.

1. The self-adjoint form and the killed form are closed on represented domains.
2. The target gate, quotient defect, and every target-sensitive operator term are represented by the affordable observable algebra or by a paid description charged to $\mathcal B$.
3. The gauge space $\mathcal G$ is closed, quotient-compatible, and has zero affordable target readout.
4. Markov capacity and polarity rows are used only when the form is regular or quasi-regular Markovian and the associated right process/capacity theory exists.
5. Conservative, unitary, or dispersive rows are used only when the corresponding commutator, carrier, passivity, or separator data are represented on a common core and descend to the propagated quotient.
6. Pseudospectral rows are used only when the closed operator $A$, the resolvent window $\Omega_A$, and the affordable input and target-output projections are represented, and the compressed resolvent $P_{\Sigma,\mathcal B}(\lambda I-A)^{-1}P_{\mathcal B}$ is computed or bounded on that window.
7. Every component outside the affordable target algebra and outside the paid description ledger is assigned to $J_{\rm res}^{0\Sigma}$ for budgeted target progress, unless an affordable target witness is supplied; in that case the component is not residual and the witness is charged to $\mathcal B$.

### Theorem 1.2 (Hamiltonian Registration)
A legal Hamiltonian signature registers in the continuum interface row by row. The closed form $\mathfrak q_H$ supplies the symmetric Geom form, $H$ or $K_\Sigma$ supplies the defect geometry for spectral/resistance rows, a represented non-normal $A$ supplies the compressed resolvent row when declared, represented target gates supply the target readout, and $\mathcal G$ supplies the initial gauge ledger. A continuous fast-track theorem applies to the Hamiltonian signature exactly when the local hypotheses for that theorem are included in the Hamiltonian signature data.

**Proof.** The representation theorem for nonnegative self-adjoint operators gives a closed nonnegative quadratic form and functional calculus. Hence the spectral measure, resolvent, pseudoinverse, and effective-resistance rows are available for represented defects in the form-dual space. If a represented closed non-normal $A$ and a resolvent window are supplied, the bounded resolvent $(\lambda I-A)^{-1}$ gives the compressed target-active pseudospectral row of Section 3b. If the form is regular or quasi-regular Markovian, the Beurling-Deny/quasi-regular theory supplies the right process, capacity, and polar-set conclusions used by the continuous notebook. If the Hamiltonian is used in a conservative or dispersive row, the supplied common core and commutator/carrier data are precisely the extra hypotheses required by the commutator and flux fast-tracks. Gauge directions have zero affordable target readout and are placed in the closed ledger. Conditional expectations register as gauge/readout projections on the GNS Hilbert space, so their ranges are represented structure and their orthogonal complements are residual for all represented readouts in the subalgebra. Expectation gauges register as exact gauges of the annealed action together with a separate quenched fluctuation-resistance row. The affordable-algebra clause is the Hamiltonian form of affordable target-completeness: target progress is measured only by represented affordable target functionals, while components invisible to all of them lie in $J_{\rm res}^{0\Sigma}$ unless a paid target witness is added. Therefore the Hamiltonian data satisfy exactly the continuum rows whose hypotheses are supplied, and no stronger row is imported without its local certificate. $\square$


## 2. Computation as a Hamiltonian

### Definition 2.1 (Computation Generator)
A Turing machine $M$ on inputs of size $n$, run under resource envelope $\mathcal B(n)$, has finite configuration space $\Omega_n$ and transition kernel $P_n$. The continuous-time generator is
$$
L_n=\rho(P_n-I),\qquad \rho>0.
$$
The reference measure $\mu_n$ is fixed by the presentation, not by a solved trajectory, and the computation Hilbert space is $\mathcal H_n=L^2(\Omega_n,\mu_n)$.


### Theorem 2.2 (Poissonized Markov-Operator Bridge)
Let $P_n$ be the discrete transition kernel of a resource-bounded machine and $L_n=\rho(P_n-I)$. Then
$$
e^{tL_n}=e^{-\rho t}\sum_{m\ge0}\frac{(\rho t)^m}{m!}P_n^m.
$$
Thus continuous time samples the same discrete computation after a Poisson number of steps. Polynomial discrete time and polynomial continuous time are equivalent up to the fixed rate $\rho$, and target-projected resolvent estimates for $L_n$ are faithful spectral stopwatches for the original machine.

**Proof.** Since $P_n$ commutes with $I$, exponentiation gives
$$
e^{t\rho(P_n-I)}=e^{-\rho t}e^{\rho tP_n}=e^{-\rho t}\sum_{m\ge0}\frac{(\rho t)^m}{m!}P_n^m.
$$
The coefficients are the probabilities of a Poisson random variable $N_t$ with mean $\rho t$. Standard Poisson concentration shows that $N_t$ is within constant factors of $\rho t$ with exponentially small tails, so polynomial step horizons correspond to polynomial continuous times. Taking the Laplace transform of the target hit flux gives the resolvent, hence the target-projected resolvent measures the same arrival process. $\square$

### Definition 2.3 (Computation Hamiltonian)
Decompose the finite generator into symmetric and antisymmetric parts in $L^2(\Omega_n,\mu_n)$:
$$
L_n=L_n^{\mathsf s}+L_n^{\mathsf a},\qquad (L_n^{\mathsf s})^*=L_n^{\mathsf s},\qquad (L_n^{\mathsf a})^*=-L_n^{\mathsf a}.
$$
The computation Hamiltonian is
$$
H_n=-L_n^{\mathsf s}.
$$
Equivalently,
$$
\langle f,H_ng\rangle_{\mu_n}=\frac12\sum_{x,z}c_n(x,z)(f(x)-f(z))(g(x)-g(z)),
$$
with nonnegative symmetric conductances $c_n$.

### Theorem 2.4 ($H_n$ is a Hamiltonian)
$H_n$ is self-adjoint and nonnegative on the finite-dimensional Hilbert space $\mathcal H_n$, hence has an ordered real spectrum
$$
0\le E_0\le E_1\le\cdots\le E_{|\Omega_n|-1}.
$$

**Proof.** The displayed Dirichlet form is symmetric and nonnegative. In finite dimension every symmetric nonnegative form is represented by a unique self-adjoint nonnegative matrix. The finite-dimensional spectral theorem gives the ordered real spectrum and an orthonormal eigenbasis. $\square$

### Definition 2.5 (Target Killing and Killed Gap)
Let $T_n\subseteq\Omega_n$ be the accepting sector. The killed Hamiltonian is
$$
K_n=H_n+V_{T_n},
$$
where $V_{T_n}$ is the target killing potential or the equivalent absorbing boundary form. Its killed-target spectral gap is
$$
\gamma_T(K_n)=\lambda_{\min}(K_n|_{\mathcal H_{\rm act}}),
$$
where $\mathcal H_{\rm act}$ is the smallest reducing subspace generated by the target gate and the initial distribution. This active restriction removes spectral modes invisible to the accepting observable.

### Theorem 2.6 (Accepting Computation is the Active Ground Mode)
For every initial vector $u_0$ in the active subspace, the killed semigroup satisfies
$$
\|e^{-tK_n}u_0\|\le e^{-t\gamma_T(K_n)}\|u_0\|.
$$
If the active ground projection $P_0u_0$ is nonzero, the large-time asymptotic has leading term
$$
e^{-tK_n}u_0=e^{-t\gamma_T(K_n)}P_0u_0+O(e^{-t(\gamma_T+\eta)}),
$$
where $\eta$ is the next active spectral gap when it exists.

**Proof.** Apply the spectral theorem to $K_n$ on the active finite-dimensional subspace. The operator norm of $e^{-tK_n}$ is the maximum of $e^{-t\lambda}$ over active eigenvalues, hence $e^{-t\gamma_T}$. If the ground projection is nonzero, separating the lowest active eigenvalue from the next one gives the stated expansion. $\square$


## 3. Active Spectral Measure, Resolvent, and Resistance

The Hamiltonian stopwatch is read on the affordable target-active vector. The exact invariant is the inverse first moment of the spectral measure seen by the affordable target algebra. The scalar gap used by this framework is the normalized reciprocal of that invariant. The raw bottom of spectral support is retained only as a diagnostic when a problem asks for an ordinary operator-gap statement.

### Definition 3.1 (Target-Active Spectral Measure and Raw Support Gap)
Let $K\ge0$ be a killed self-adjoint Hamiltonian and let $E$ be a represented target defect or gate vector after quotienting kernel, gauge, and zero-target residual directions when those quotients have already been declared. The active spectral measure is
$$
\nu_E(B)=\|\mathbf 1_B(K)E\|^2
$$
for Borel sets $B\subset[0,\infty)$. Its raw support gap is the diagnostic number
$$
\gamma_{\rm raw}(K,E)=\inf\bigl(\operatorname{supp}\nu_E\cap(0,\infty)\bigr),
$$
with the convention $\gamma_{\rm raw}=0$ when the positive support has no positive lower bound. This raw support infimum is not the default framework gap.

### Definition 3.2 (Hamiltonian Resistance and Resolvent Saturation)
The Hamiltonian effective resistance of $E$ is
$$
\mathcal R_K(E)=\langle E,K^+E\rangle
=\int_{(0,\infty)}s^{-1}\,d\nu_E(s),
$$
when the integral is finite, and $+\infty$ otherwise. The discounted target resolvent saturation is
$$
\mathsf S_\lambda(E)=\langle E,(K+\lambda I)^{-1}E\rangle
=\int_{[0,\infty)}(\lambda+s)^{-1}\,d\nu_E(s).
$$
If $P_{\ker K}E=0$, this is the Laplace/Green stopwatch of the killed dynamics on the target-active channel.

### Definition 3.3 (Affordable Target-Active Transcript and Framework Gap)
Fix a budget $\mathcal B$ and a propagated closed ledger $\mathcal K_{\rm closed}^{\Sigma}$. Let $\mathcal A_{\mathcal B}^{\Sigma}$ be the closed subspace of the reduced Hilbert space generated by affordable target gates, affordable target-shell functionals, and paid target defects. Let $P_{\mathcal A}$ denote orthogonal projection onto $\mathcal A_{\mathcal B}^{\Sigma}$ and define
$$
E_{\mathcal B}=P_{\mathcal A}P_{(\ker K)^\perp}E,
\qquad
E_{\rm res}=(I-P_{\mathcal A})P_{(\ker K)^\perp}E.
$$
An affordable target-active transcript for a family $(K_n,E_n)$ consists of the measures $\nu_{E_{\mathcal B,n}}$ and the following certificates:

1. $E_{{\rm res},n}$ pairs to zero with every affordable target shell and hence belongs to $J_{\rm res}^{0\Sigma}$ by affordable target-completeness. If such a pairing is nonzero, the pairing is an affordable witness and the corresponding component is moved into $\mathcal A_{\mathcal B}^{\Sigma}$ with its cost charged to $\mathcal B$.
2. When a normalized scalar certificate is read, the affordable active defect has budget-unit size:
   $$
   0<c_-\le \|E_{\mathcal B,n}\|\le c_+<\infty
   $$
   after the declared normalization. If $E_{\mathcal B,n}=0$, the framework gap below is defined to be zero and no positive budgeted target progress is certified.

The framework, or stopwatch, gap is
$$
\gamma_{\mathcal B}^{\Sigma}(K,E)
=
\begin{cases}
0,& \|E_{\mathcal B}\|=0\text{ or }\mathcal R_K(E_{\mathcal B})=+\infty,\\[2mm]
\dfrac{\|E_{\mathcal B}\|^2}{\mathcal R_K(E_{\mathcal B})},&0<\mathcal R_K(E_{\mathcal B})<+\infty.
\end{cases}
$$
Equivalently, $\gamma_{\mathcal B}^{\Sigma}$ is the resistance gap of the affordable target-active vector. Only $\nu_{E_{\mathcal B,n}}$ is used for budgeted target progress. The residual complement is closed before any gap, resistance, flux, or capacity verdict is read.

### Theorem 3.4 (Resolvent-Resistance-Stopwatch Identity)
Assume $E$ has no component in $\ker K$ after gauge and residual zero-target directions have been quotient out. Then
$$
\lim_{\lambda\downarrow0}\mathsf S_\lambda(E)=\mathcal R_K(E).
$$
For the affordable transcript of Definition 3.3,
$$
\gamma_{\mathcal B}^{\Sigma}(K,E)
=\frac{\|E_{\mathcal B}\|^2}{\lim_{\lambda\downarrow0}\mathsf S_\lambda(E_{\mathcal B})}
$$
with the same zero/infinite conventions. Consequently, under the normalization in Definition 3.3, the following are equivalent in any declared budget class closed under constant-factor comparison:
$$
\mathcal R_K(E_{\mathcal B,n})\preceq b(n)
\qquad\Longleftrightarrow\qquad
\gamma_{\mathcal B}^{\Sigma}(K_n,E_n)\succeq b(n)^{-1}.
$$
Moreover,
$$
\gamma_{\mathcal B}^{\Sigma}(K,E)\ge \gamma_{\rm raw}(K,E_{\mathcal B})
$$
whenever the raw positive support gap is positive. Equality holds when $\nu_{E_{\mathcal B}}$ is supported on its bottom active eigenspace. No converse comparison with $\gamma_{\rm raw}$ follows from resistance alone.

**Proof.** The spectral theorem gives the displayed integral formulas for $\mathsf S_\lambda$ and $\mathcal R_K$. Since the kernel component has been removed, $(\lambda+s)^{-1}$ increases pointwise to $s^{-1}$ on $(0,\infty)$. Monotone convergence gives
$$
\lim_{\lambda\downarrow0}\mathsf S_\lambda(E)=\int_{(0,\infty)}s^{-1}\,d\nu_E(s)=\mathcal R_K(E).
$$
Applying this to $E_{\mathcal B}$ and substituting Definition 3.3 gives the formula for $\gamma_{\mathcal B}^{\Sigma}$.

If $c_-\le \|E_{\mathcal B,n}\|\le c_+$, then
$$
\gamma_{\mathcal B}^{\Sigma}(K_n,E_n)
=\|E_{\mathcal B,n}\|^2/\mathcal R_{K_n}(E_{\mathcal B,n}).
$$
Thus $\mathcal R_{K_n}(E_{\mathcal B,n})\preceq b(n)$ implies $\gamma_{\mathcal B}^{\Sigma}\succeq b(n)^{-1}$, and the reverse implication follows from the lower norm bound. The zero and infinite cases are exactly the declared conventions.

If $\gamma_{\rm raw}=a>0$, the support of $\nu_{E_{\mathcal B}}$ lies in $[a,\infty)$ on the positive spectral subspace. Hence
$$
\mathcal R_K(E_{\mathcal B})
=\int s^{-1}\,d\nu_{E_{\mathcal B}}(s)
\le a^{-1}\|E_{\mathcal B}\|^2,
$$
which gives $\gamma_{\mathcal B}^{\Sigma}\ge a$. If the measure is supported at $a$, equality is immediate. The converse fails in general: a measure may place arbitrarily small but nonzero mass near zero, making $\gamma_{\rm raw}=0$, while its inverse first moment remains budget-admissible. Therefore raw support information is an additional diagnostic, not part of the stopwatch identity. $\square$

### Corollary 3.5 (Affordable Reachability, Resolvent, and Framework Gap are the Same Invariant)
For a target-active family with an affordable target-active transcript in the sense of Definition 3.3, the following are equivalent in the declared budget class:

1. the affordable target defect is routable at budget $b(n)$;
2. the zero-discount affordable target resolvent saturation is bounded by $b(n)$;
3. the Hamiltonian resistance $\mathcal R_{K_n}(E_{\mathcal B,n})$ is bounded by $b(n)$;
4. the framework gap $\gamma_{\mathcal B}^{\Sigma}(K_n,E_n)$ is lower-bounded at reciprocal budget scale.

In this equivalence, $E_{{\rm res},n}$ cannot create a budgeted target gap or budgeted target arrival: it is either zero-target residual by the target-completeness theorem, or it has a nonzero affordable target pairing and has already been moved into $E_{\mathcal B,n}$ with its cost charged.

**Proof.** Items 2 and 3 are identical by Theorem 3.4. Items 3 and 4 are identical up to the declared norm normalization by the definition of $\gamma_{\mathcal B}^{\Sigma}$. Item 1 is the routing predicate of Definition 4.1, namely bounded affordable resistance on the reduced target-active vector. It remains only to justify that no discarded component can change the verdict. Definition 3.3 and affordable target-completeness place $E_{{\rm res},n}$ in $J_{\rm res}^{0\Sigma}$, so it pairs trivially with every affordable target shell and contributes no budgeted target progress. Paper 3's arrival-to-visibility accountability theorem says conversely that target arrival above residual baseline by an affordable computation supplies counted structural visibility in the same budget ledger. Such a counted target-visible component is precisely part of $\mathcal A_{\mathcal B}^{\Sigma}$ and hence belongs to $E_{\mathcal B,n}$. Therefore the affordable target projection exhausts all budget-relevant target progress, and the four displayed conditions are the same invariant. $\square$

## 3b. Target-Active Pseudospectral Stopwatch

Self-adjoint Hamiltonian rows are measured by a spectral measure. Non-normal rows are measured by the same stopwatch through the target-active resolvent. The framework never uses the ambient pseudospectrum as a certificate. It uses only the part of the resolvent seen from represented affordable inputs and represented affordable target readouts.

### Definition 3b.1 (Affordable Target-Active Pseudospectral Transcript)
Let $A$ be a closed densely defined operator on the reduced Hilbert space after the declared kernel, gauge, and closed-ledger quotients have been applied. A pseudospectral transcript consists of:

1. a represented resolvent window $\Omega_A\subset\rho(A)$;
2. closed subspaces $\mathcal H_{\mathcal B}$ and $\mathcal H_{\Sigma,\mathcal B}$ generated by affordable input/defect representatives and affordable target-output representatives;
3. the orthogonal projections $P_{\mathcal B}$ and $P_{\Sigma,\mathcal B}$ onto these subspaces;
4. budget-dense represented subsets $\mathcal D_{\mathcal B}\subset\mathcal H_{\mathcal B}$ and $\mathcal D_{\Sigma,\mathcal B}\subset\mathcal H_{\Sigma,\mathcal B}$ whose finite representatives and precisions are charged to $\mathcal B$.

For $\lambda\in\Omega_A$, the target-active pseudospectral resolvent is
$$
T_{\lambda,\mathcal B}^{\Sigma}(A)
= P_{\Sigma,\mathcal B}(\lambda I-A)^{-1}P_{\mathcal B}.
$$
Its pseudospectral stopwatch size is
$$
\Psi_{\mathcal B}^{\Sigma}(\lambda;A)
=\|T_{\lambda,\mathcal B}^{\Sigma}(A)\|,
$$
and the reciprocal pseudospectral stopwatch gap is
$$
\gamma_{{\rm ps},\mathcal B}^{\Sigma}(\lambda;A)=
\begin{cases}
+\infty,& \Psi_{\mathcal B}^{\Sigma}(\lambda;A)=0,\\[1mm]
\Psi_{\mathcal B}^{\Sigma}(\lambda;A)^{-1},&0<\Psi_{\mathcal B}^{\Sigma}(\lambda;A)<+\infty,\\[1mm]
0,& \Psi_{\mathcal B}^{\Sigma}(\lambda;A)=+\infty.
\end{cases}
$$
For represented unit vectors $f\in\mathcal D_{\mathcal B}$ and $g\in\mathcal D_{\Sigma,\mathcal B}$, the scalar transcript is
$$
\psi_\lambda(g,f;A)=\langle g,(\lambda I-A)^{-1}f\rangle.
$$
The target-active $\epsilon$-pseudospectrum is the set of represented resolvent points $\lambda\in\Omega_A$ where $\Psi_{\mathcal B}^{\Sigma}(\lambda;A)>\epsilon^{-1}$. Ambient resolvent growth outside $P_{\mathcal B}$, invisible to $P_{\Sigma,\mathcal B}$, or lacking represented witnesses is not a target-progress certificate.

### Theorem 3b.2 (Pseudospectral Duality and Stopwatch Representation)
Let $A$ be closed and densely defined, let $\lambda\in\Omega_A\subset\rho(A)$, and let the transcript of Definition 3b.1 be supplied. Then $T_{\lambda,\mathcal B}^{\Sigma}(A)$ is bounded and
$$
\Psi_{\mathcal B}^{\Sigma}(\lambda;A)
=
\sup_{\substack{\|f\|=1,
\ f\in \mathcal H_{\mathcal B}\\
\|g\|=1,
\ g\in \mathcal H_{\Sigma,\mathcal B}}}
|\langle g,(\lambda I-A)^{-1}f\rangle|,
$$
with the convention that the supremum is $0$ if either projected subspace is zero. The same supremum is obtained by restricting $f$ and $g$ to normalized nonzero elements of the budget-dense represented sets $\mathcal D_{\mathcal B}$ and $\mathcal D_{\Sigma,\mathcal B}$. Consequently, for every $0<\eta<\Psi_{\mathcal B}^{\Sigma}(\lambda;A)$ there are represented affordable unit witnesses $f$ and $g$ such that
$$
|\psi_\lambda(g,f;A)|>\eta.
$$
If $A$ generates a $C_0$ semigroup $(e^{tA})_{t\ge0}$ with growth bound $\omega$ and $\operatorname{Re}\lambda>\omega$, then
$$
\psi_\lambda(g,f;A)=
\int_0^\infty e^{-\lambda t}\langle g,e^{tA}f\rangle\,dt.
$$
Thus the target-active pseudospectral resolvent is exactly the discounted target stopwatch for non-normal dynamics.

**Proof.** Since $\lambda\in\rho(A)$ and $A$ is closed, the closed graph theorem gives a bounded resolvent $(\lambda I-A)^{-1}$ on the whole Hilbert space. Hence $T_{\lambda,\mathcal B}^{\Sigma}(A)$ is bounded. If either projected subspace is zero, then $T_{\lambda,\mathcal B}^{\Sigma}=0$ and the displayed identity holds by convention. Otherwise the Hilbert-space norm identity gives
$$
\|T\|=\sup_{\|f\|=1}\|Tf\|
=\sup_{\|f\|=1}\sup_{\|g\|=1}|\langle g,Tf\rangle|.
$$
Since $T=P_{\Sigma,\mathcal B}(\lambda I-A)^{-1}P_{\mathcal B}$, the first supremum may be restricted to $\mathcal H_{\mathcal B}$ and the second to $\mathcal H_{\Sigma,\mathcal B}$. The dense-set version follows by continuity of the bounded bilinear form
$$
(g,f)\mapsto \langle g,(\lambda I-A)^{-1}f\rangle
$$
on the product of the two Hilbert subspaces: approximate any unit pair by nonzero represented vectors from the dense sets and renormalize. This gives represented approximate witnesses for every $\eta<\Psi_{\mathcal B}^{\Sigma}$. If $A$ generates a $C_0$ semigroup and $\operatorname{Re}\lambda>\omega$, the Hille-Yosida resolvent formula gives
$$
(\lambda I-A)^{-1}f=
\int_0^\infty e^{-\lambda t}e^{tA}f\,dt
$$
as a Bochner integral. Pairing with $g$ gives the scalar stopwatch identity. $\square$

### Theorem 3b.3 (Pseudospectral Amplification Has No Fourth Channel)
Assume a legal Hamiltonian or generator interface, an affordable target compactification, a target-complete propagated ledger, and a represented pseudospectral transcript in the sense of Definition 3b.1. Let $A$ be the represented non-normal row and let $\lambda\in\Omega_A$ be a represented discount/resolvent point. Every budget-admissible target-active pseudospectral amplification detected by $\Psi_{\mathcal B}^{\Sigma}(\lambda;A)$ is exhausted by the existing channel taxonomy:

1. it is generated by public or paid represented structure and is charged to the affordable algebra, with source in $\mathsf{Geom}$ or $\mathsf{Abs}$ and with $\mathsf{Caus}$, $\mathsf{Lift}$, or $\mathsf{Bdry}$ only as legal uses or readouts of that source; or
2. it is invisible to every affordable target shell and lies in $J_{\rm res}^{0\Sigma}$, hence it contributes no budgeted target progress.

A residual-looking component with a represented nonzero pairing against an affordable target shell is not residual. The scalar transcript witnessing that pairing is an affordable target witness; the component is moved into the paid/structural ledger with its representation cost charged. A target-visible unclassified pseudospectral witness is therefore an incomplete Abs/Geom/provenance row, not a fourth kind of mechanism.

**Proof.** Suppose $\Psi_{\mathcal B}^{\Sigma}(\lambda;A)>\eta>0$. By Theorem 3b.2 there are represented affordable unit witnesses $f\in\mathcal D_{\mathcal B}$ and $g\in\mathcal D_{\Sigma,\mathcal B}$ with
$$
|\langle g,(\lambda I-A)^{-1}f\rangle|>\eta.
$$
Let $u=(\lambda I-A)^{-1}f$. Then $u\in\mathcal D(A)$ and
$$
(\lambda I-A)u=f.
$$
The vector $u$ need not be an affordable primitive. The budgeted witness is the represented transcript consisting of $A$, $\lambda$, $f$, $g$, and the scalar resolvent readout; if a proof uses $u$ itself as data, the representation cost of $u$ is charged.

Apply the framework provenance partition to the operator row, the input defect, and the target readout. A represented term generated by the public problem signature belongs to the exposed structural ledger. Its intrinsic source is $\mathsf{Geom}$ or $\mathsf{Abs}$; directed movement, lifted coordinates, and boundary readouts are the $\mathsf{Caus}$, $\mathsf{Lift}$, and $\mathsf{Bdry}$ uses of those sources. A represented term not public but present in the operator, advice, learned geometry, finite transcript, or represented resolvent witness is paid description and is charged to $\mathcal B$. These are the structural alternatives.

If the component is neither public nor paid, it is residual by provenance exhaustiveness. Target-completeness now applies. When it pairs trivially with every affordable target shell, it lies in $J_{\rm res}^{0\Sigma}$ and cannot count as budgeted target progress. When it has a represented nonzero affordable target pairing, that pairing is precisely the target witness above, so the component is reclassified into the affordable target ledger with cost charged. Hence no target-visible residual component remains residual. The two displayed alternatives exhaust all budget-admissible cases. $\square$

### Corollary 3b.4 (Self-Adjoint Reduction)
Let $K\ge0$ be self-adjoint and set $A=-K$. For $\lambda>0$ and any affordable target vector $E_{\mathcal B}$,
$$
\langle E_{\mathcal B},(\lambda I-A)^{-1}E_{\mathcal B}\rangle
=\langle E_{\mathcal B},(K+\lambda I)^{-1}E_{\mathcal B}\rangle
=\mathsf S_\lambda(E_{\mathcal B}).
$$
If $E_{\mathcal B}\ne0$ and $P_{\mathcal B}=P_{\Sigma,\mathcal B}$ is the rank-one projection onto $e=E_{\mathcal B}/\|E_{\mathcal B}\|$, then
$$
\Psi_{\mathcal B}^{\Sigma}(\lambda;-K)
=\langle e,(K+\lambda I)^{-1}e\rangle
=\mathsf S_\lambda(E_{\mathcal B})/\|E_{\mathcal B}\|^2.
$$
If $E_{\mathcal B}=0$, both the rank-one projected stopwatch and the framework target progress are zero by convention. Consequently the pseudospectral stopwatch reduces to the spectral stopwatch of Section 3, and the zero-discount limit recovers the resistance gap whenever Theorem 3.4 applies.

**Proof.** Since $A=-K$, $(\lambda I-A)^{-1}=(K+\lambda I)^{-1}$. This proves the first equality. If $E_{\mathcal B}\ne0$, the rank-one compression to the unit vector $e$ is multiplication by the nonnegative scalar $\langle e,(K+\lambda I)^{-1}e\rangle$, so its operator norm is the displayed scalar. The zero-vector case follows from the projection convention. The limit statement is Theorem 3.4 applied to $E_{\mathcal B}$. $\square$

### Theorem 3b.5 (Pseudospectral Fast-Track Equivalence)
For a registered non-normal row with a represented pseudospectral transcript, an explicit calculation of transient growth, Jordan amplification, non-normal mode coupling, or resolvent blow-up gives the same budget verdict as the fast-track invariant $\Psi_{\mathcal B}^{\Sigma}$ only after it has been converted, on the same affordable projections and closed-ledger quotient, into an upper or lower estimate for the compressed resolvent. At a specified discount $\lambda(n)$, a successful bound
$$
\Psi_{\mathcal B}^{\Sigma}(\lambda(n);A_n)\preceq b(n)
$$
closes the pseudospectral row at budget $b(n)$; on a declared window $\Omega_{A,n}$, the corresponding closure statement is the uniform bound
$$
\sup_{\lambda\in\Omega_{A,n}}\Psi_{\mathcal B}^{\Sigma}(\lambda;A_n)\preceq b(n).
$$
A lower-bound or failure certificate consists of represented affordable witnesses $f_n,g_n$ with over-budget target-active resolvent response; by Theorem 3b.3 those witnesses must be classified as Abs/Geom structure, charged paid description, or target-invisible $J_{\rm res}^{0\Sigma}$. A large ambient pseudospectrum without represented target-active witnesses is only an ambient diagnostic and gives no budgeted framework verdict.

**Proof.** After the explicit non-normal calculation has been converted into the compressed resolvent row, it is an upper or lower bound on
$$
\sup_{\substack{\|f\|=1,
\ f\in\mathcal H_{\mathcal B}\\
\|g\|=1,
\ g\in\mathcal H_{\Sigma,\mathcal B}}}
|\langle g,(\lambda I-A)^{-1}f\rangle|.
$$
By Theorem 3b.2 this supremum is exactly $\Psi_{\mathcal B}^{\Sigma}$. Therefore an explicit upper bound and a fast-track upper bound are the same statement, pointwise in $\lambda$ or uniformly after taking the displayed supremum over the represented window. If the bound fails through a represented lower-bound transcript, Theorem 3b.2 gives represented approximate unit witnesses with the corresponding large response. Theorem 3b.3 classifies those witnesses through the existing structural exhaustion. If the only large quantity is an unprojected or unrepresented ambient norm, the hypotheses of the pseudospectral row are not met, so no budgeted target-progress conclusion follows. Hence the fast-track and the explicit represented pseudospectral calculation have identical closed/open verdicts. $\square$


## 4. Routable Versus Hard Families

### Definition 4.1 (Routable Hamiltonian Family)
A Hamiltonian family $(K_n,E_n)$ is routable at budget $\mathcal B$ when
$$
\mathcal R_{K_n}(E_{\mathcal B,n})\preceq b(n)
$$
for the affordable target-active vector supplied by Definition 3.3, after quotienting by gauge, zero-target residual classes, and any propagated closed ledger. Equivalently, the full represented defect is routed at budget after its non-affordable target-invisible complement has been placed in $J_{\rm res}^{0\Sigma}$.

### Theorem 4.2 (The Framework Returns a Verdict-Bearing Gap Certificate)
For every registered Hamiltonian family with an affordable target-active transcript, exactly one of the following holds inside the declared budget class.

- **Routable.** $\mathcal R_{K_n}(E_{\mathcal B,n})$ is budget-admissible. The framework returns the affordable resistance and the framework/stopwatch gap $\gamma_{\mathcal B}^{\Sigma}(K_n,E_n)$.
- **Non-routable.** $\mathcal R_{K_n}(E_{\mathcal B,n})$ is not budget-admissible. The framework returns the divergent inverse-moment certificate; equivalently $\gamma_{\mathcal B}^{\Sigma}(K_n,E_n)$ closes relative to the declared budget class.

**Proof.** Routability is the predicate $\mathcal R_{K_n}(E_{\mathcal B,n})\preceq b(n)$. Its negation is non-routability relative to the declared budget. The residual complement is already in $J_{\rm res}^{0\Sigma}$ by Definition 3.3, so it cannot contribute to budgeted target progress. The resistance is the exact spectral inverse moment by Theorem 3.4 applied to $E_{\mathcal B,n}$, and Corollary 3.5 identifies it with the framework gap. These alternatives are mutually exclusive and exhaustive for the represented budget class. $\square$

### Theorem 4.3 (No Universal Affordable Framework-Gap Estimator)
There is no family-uniform affordable procedure that returns a constant or inverse-budget framework/stopwatch gap for every computation Hamiltonian family. Equivalently, no universal affordable routing closes every affordable target defect.

**Proof.** Suppose such a procedure existed. By Corollary 3.5 it would give a family-uniform budget-admissible bound on $\mathcal R_{K_n}(E_{\mathcal B,n})$ for every registered family, hence every affordable target defect would be routable at that budget. That is exactly the negation of the framework's saturated-floor theorem and its seal families: those families have no exposed structural channel at the declared budget after all accountable channels have been removed. Therefore the assumed universal estimator cannot exist. $\square$

### Theorem 4.4 (Routable/Non-Routable Dichotomy is Exhaustive)
A registered computation family is either routable or non-routable relative to $\mathcal B$ on the affordable target-active subspace. With the affordable target-active transcript supplied, this is equivalently a reciprocal-budget framework gap or a framework gap that closes faster than the declared reciprocal budget. There is no third outcome inside the budget class.

**Proof.** The budget order decides whether $\mathcal R_{K_n}(E_{\mathcal B,n})\preceq b(n)$. The negation is failure for that budget class. Corollary 3.5 transports the resistance dichotomy to $\gamma_{\mathcal B}^{\Sigma}$. The complement of $E_{\mathcal B,n}$ is residual zero-target for the affordable algebra, unless an affordable witness is added and charged; hence it does not define a third budgeted target-progress outcome. $\square$


## 5. Continuous Hamiltonians and PDE

### Definition 5.1 (Continuous Hamiltonian Interface)
A continuous Hamiltonian interface consists of a legal Hamiltonian signature in which $H$ is the self-adjoint operator associated with a regular or quasi-regular Dirichlet form, a Schrodinger form, a wave/Klein-Gordon energy form, or another closed positive energy form. The target $\Sigma$ is represented by a capacity, a killing potential, an absorbing boundary, or a target shell family.

### Theorem 5.2 (Killed Continuous Hamiltonian and Capacity)
The killed continuous Hamiltonian $K_\Sigma$ exists as a nonnegative self-adjoint operator whenever the killed form is closed. If the Hamiltonian form is regular or quasi-regular Markovian and
$$
{\rm Cap}_{\mathcal B}(\Sigma)=0,
$$
then $\Sigma$ is polar for the associated affordable right process and cannot be reached by finite-energy affordable states. For a non-Markovian conservative Hamiltonian, the same zero-capacity statement is a ledger exclusion inside the affordable target algebra: no budgeted finite-energy target functional can concentrate on $\Sigma$, but a probabilistic hitting claim requires a supplied right-process or nonpolarity interface.

**Proof.** Closedness of the killed form gives $K_\Sigma$ by the Friedrichs representation theorem. For regular and quasi-regular Markovian forms, the continuous notebook's zero-capacity theorem applies: capacity-zero sets are exceptional for the associated right process and removable for finite-energy quasi-continuous representatives. In a purely conservative Hamiltonian row there need not be a right process. The capacity object still defines a closed zero-capacity ledger class for affordable target functionals, so it excludes budgeted target progress in the affordable algebra; it does not assert stochastic hitting without the Markov interface. $\square$

### Theorem 5.3 (Continuous Spectral Routing)
Let $E_{\mathcal B,\Sigma}$ be the affordable target-active projection of the represented target defect or gate for a continuous Hamiltonian. The budgeted routing certificate is
$$
\mathcal R_{K_\Sigma}(E_{\mathcal B,\Sigma})=\langle E_{\mathcal B,\Sigma},K_\Sigma^+E_{\mathcal B,\Sigma}\rangle.
$$
If this is finite after quotienting by the propagated closed ledger, the affordable target defect is routed. If it is infinite, the non-routable component is the kernel/zero-spectrum or inverse-moment obstruction of the affordable active spectral measure and must appear as a named open quotient unless it has zero affordable target flux or zero target capacity. The complement of $E_{\mathcal B,\Sigma}$ is already $J_{\rm res}^{0\Sigma}$ for budgeted target progress by Definition 3.3.

**Proof.** This is Theorem 3.4 applied to the killed continuous Hamiltonian and the affordable active vector. The continuous notebook's target-completeness theorem places every zero-flux, zero-capacity, gauge, routed, dyadic, rebased, subthreshold, or separator-killed class in the closed ledger, and Definition 3.3 places the non-affordable target-invisible complement in $J_{\rm res}^{0\Sigma}$. Therefore an infinite affordable resistance that survives the quotient is exactly the open spectral obstruction. $\square$

### Theorem 5.4 (Boundary Defects, Constraint Tails, and Spectral Routing Commute)
Assume the continuous fast-track rows for boundary defects, constraint tails, and ledger propagation have succeeded before the Hamiltonian routing row is computed. Then the Hamiltonian resistance, affordable target-active spectral measure, and framework/stopwatch gap computed on the reduced space are canonically the same as the unreduced spectral objects modulo all already closed classes and the $J_{\rm res}^{0\Sigma}$ complement.

**Proof.** The propagated ledger is a closed subspace or closed cone and the Hamiltonian form descends to the quotient by the quotient-compatibility hypothesis. The third isomorphism theorem for closed ledgers identifies later reduced quotients with unreduced quotients modulo the closed classes. Functional calculus on the descended self-adjoint form gives the same spectral measure for the reduced defect as the unreduced measure after projecting away closed classes. $\square$


## 6. Degenerate, Hypocoercive, and Dispersive Hamiltonians

### Definition 6.1 (Hamiltonian Commutator Routing Interface)
When the direct Geom operator is degenerate, the Hamiltonian signature supplies either:

1. a hypocoercive bracket tower $\{\operatorname{ad}_{\mathsf A}^k\mathcal L_W\}_{k\le K}$ and gauge $\mathcal G$; or
2. a conservative Hamiltonian $H$, a conjugate operator $A$, spectral subspace $P$, and Mourre bound $P i[H,A]P\ge\theta P$ modulo gauge.

### Theorem 6.2 (Commutator Routing is the Hamiltonian Replacement for Direct Resistance)
If the hypocoercive bracket closure is coercive modulo gauge, then routing uses
$$
\langle E,(\overline{\mathcal L_W}^{\,\mathrm{ad},K})^+E\rangle.
$$
If the system is conservative and a Mourre bound holds modulo gauge, routing uses the monotone carrier
$$
\frac{d}{dt}\langle u,Au\rangle=\langle u,i[H,A]u\rangle\ge\theta\|Pu\|^2.
$$
In either case the direct infinite resistance is replaced by a represented commutator fast-track.

**Proof.** The hypocoercive statement is the bracket-rank theorem from the continuous notebook: coercivity of the positive bracket closure gives a bounded pseudoinverse modulo gauge. The conservative statement is Heisenberg's identity plus the positive commutator estimate. Both are quotient-compatible with the Noether ledger, so the routing verdict descends to the reduced Hamiltonian space. $\square$

### Theorem 6.3 (Dispersive PDE Dichotomy)
For conservative dispersive PDE registered in the Hamiltonian interface, exactly one alternative holds.

- If the Mourre/Morawetz deficit vanishes modulo gauge, the monotone carrier gives the spacetime estimate needed to bar concentration on the target shell.
- If the deficit is nonzero, the surviving class is a resonance, standing wave, soliton, or conserved hierarchy class. It reaches the target exactly when its affordable target flux and capacity are nonzero; otherwise it is residual or zero-capacity and is closed by the ledger.

**Proof.** Vanishing deficit is the positive-commutator certificate of Theorem 6.2, hence the target shell has integrable passage density. Nonzero deficit is, by definition, a spectral subspace on which no positive commutator is available after gauge is removed. Affordable target completeness then gives the flux/capacity dichotomy for that class. $\square$


## 7. Hamiltonian Conservative Fast-Tracks

The Hamiltonian formulation must also inherit the conservative fast-track from the continuous notebook. In Hamiltonian language the carrier is an energy, action, virial, entropy, or Noether moment; the obstruction is measured only after all previously closed channels have been quotient out.

### Definition 7.1 (Hamiltonian Carrier/Passivity Interface)
A Hamiltonian carrier/passivity interface consists of:

1. a propagated closed ledger $\mathcal K_i^\Sigma$ and reduced profile space $\mathcal P_H^{\rm red}$;
2. a nonnegative represented budget/dissipation form $D_H$ on $\mathcal P_H^{\rm red}$;
3. a represented positive target-flux form $N_H^+$ on the same reduced space;
4. quotient-compatible transports from boundary, tail, gauge, residual, commutator-routed, and zero-capacity classes into $\mathcal K_i^\Sigma$.

The Hamiltonian passivity radius is
$$
\beta_H^\Sigma=\sup_{D_H(Z)>0}\frac{N_H^+(Z)}{D_H(Z)},
$$
with $D_H$-null positive-flux modes represented as a separate target-complete quotient before the radius is evaluated.

### Theorem 7.2 (Hamiltonian Flux-Sign Fast-Track)
An explicit Hamiltonian sign, virial, multiplier, or energy-flux proof of
$$
N_H^+(Z)\le D_H(Z)\qquad (Z\in\mathcal P_H^{\rm red})
$$
is equivalent to the fast-track certificate $\beta_H^\Sigma\le1$. A strict proof with margin is equivalent to $\beta_H^\Sigma<1$. If $\beta_H^\Sigma>1$, the failed fast-track gives a feeding profile or compactified feeding sequence.

**Proof.** The radius is the sharp supremum of the ratio $N_H^+/D_H$ on the reduced represented profile space. If the explicit proof holds pointwise, the supremum is at most one. If the supremum is at most one, the pointwise inequality follows by definition for every represented profile. Strictness is the same statement with the optimized margin. If the supremum is larger than one, there is a represented profile or an approximating compactified sequence whose positive target flux exceeds the dissipation budget. $\square$

### Definition 7.3 (Hamiltonian Noether-Flux Certificate)
A Noether-flux certificate is a represented symmetry algebra $\mathfrak g$, moment map $\mathcal J:\mathfrak g\to\mathcal H^*$, and coefficients $c_a$ such that the reduced target flux factors as
$$
N_H^+(Z)=\sum_a c_a\,\mathcal J(\xi_a)(Z)+N_{\rm closed}(Z),
$$
where $N_{\rm closed}$ vanishes in the quotient by the propagated ledger.

### Theorem 7.4 (Noether Flux Computes the Passivity Radius)
If a Noether-flux certificate exists, then $\beta_H^\Sigma$ is computed on the finite or represented Noether quotient obtained by projecting the moment map to $\mathcal P_H^{\rm red}/\ker D_H$. In particular, every symmetry direction in the kernel of the projected moment map is a gauge or zero-flux residual direction and belongs to the closed ledger.

**Proof.** The closed term disappears after projection to the reduced quotient. The remaining flux is a represented linear combination of Noether moments, so optimizing $N_H^+/D_H$ is exactly the operator norm or numerical radius of that moment-map functional relative to $D_H$. If a symmetry direction has zero projected moment and zero target readout, target-completeness places it in the gauge/residual ledger. $\square$

### Definition 7.5 (Hamiltonian Monotone-Entropy Alternative)
A monotone-entropy alternative is a represented functional $\mathcal M$ and nonnegative density $C_H$ such that along admissible Hamiltonian or lifted Hamiltonian trajectories,
$$
\frac{d}{dt}\mathcal M(u(t))\le -C_H(u(t))
$$
or, for dispersive/Morawetz conventions,
$$
\frac{d}{dt}\mathcal M(u(t))\ge C_H(u(t)),
$$
with the sign fixed once and with $C_H$ dominating the target passage density on the reduced quotient.

### Theorem 7.6 (Monotone Entropy Bypasses Flux-Sign Calculation)
If the monotone-entropy alternative holds, $\mathcal M$ has budget-bounded endpoint oscillation on the admissible orbit, and $C_H$ dominates the target passage density after ledger reduction, then the Hamiltonian carrier closes the reaching branch without separately computing $\beta_H^\Sigma$.

**Proof.** Integrating the monotonicity identity gives
$$
\int C_H(u(t))\,dt\le |\mathcal M(u(t_1))-\mathcal M(u(t_0))|
$$
with the inequality direction adjusted to the declared sign convention. The endpoint oscillation is budget-bounded by hypothesis. Since target passage is dominated by $C_H$, the target shell receives finite total passage density and cannot support a persistent reaching profile unless a nonzero open quotient survives. Such a survivor is precisely the named failure alternative in the target-complete ledger. $\square$


## 8. Hamiltonian Saturation and Rigidity

### Definition 8.1 (Hamiltonian Saturation Ledger)
Assume $\beta_H^\Sigma=1$. A Hamiltonian saturation ledger is a represented decomposition
$$
D_H(Z)-N_H^+(Z)=\sum_{a=1}^N Q_a(Z),\qquad Q_a(Z)\ge0,
$$
after all boundary, tail, gauge, commutator-routing, residual, and zero-capacity classes have been propagated. The equality set is the compactified represented set where every $Q_a=0$. Let
$$
\mathscr R_{H,\rm sat}:\widehat{\mathcal E}_H\to\mathcal C_H
$$
be the remaining channel map on that equality set, and define the Hamiltonian saturated quotient
$$
\mathfrak H_{H,\rm sat}^\Sigma=\overline{\Pi_\Sigma\mathscr R_{H,\rm sat}(\widehat{\mathcal E}_H)}/\mathcal K_{H,\rm closed}^\Sigma.
$$

### Theorem 8.2 (Hamiltonian Borderline Closure)
In the borderline case $\beta_H^\Sigma=1$, the equality branch closes if and only if
$$
\mathfrak H_{H,\rm sat}^\Sigma=0.
$$
If this quotient is nonzero, the nonzero class is exactly the saturated equality remainder that the explicit rigidity analysis would need to isolate.

**Proof.** Along an extremizing sequence, $D_H-N_H^+\to0$. Since the ledger defects $Q_a$ are nonnegative and sum to this difference, each $Q_a\to0$, so every compactified extremizer satisfies the saturation equations. The map $\mathscr R_{H,\rm sat}$ records what remains after all equality constraints and previously closed ledgers have been applied. Vanishing of the quotient says every such remainder lies in the closed ledger; nonvanishing gives precisely a remainder outside that ledger. $\square$

### Definition 8.3 (Hamiltonian Separator)
A Hamiltonian separator is a finite represented combination of Pohozaev, Noether, virial, Mourre, modulation, Liouville, or monotonicity identities whose projection to the open quotient gives
$$
0=\|\Pi_{\rm open}\mathscr R_{H,\rm sat}(Z)\|_{\mathcal Q}^2+\sum_j q_j(Z)+b_{\rm closed}(Z),
$$
where $q_j\ge0$ and $b_{\rm closed}$ vanishes in the quotient.

### Theorem 8.4 (Separators Force Saturation Vanishing)
If a Hamiltonian separator is positive definite on the open quotient, then
$$
\mathfrak H_{H,\rm sat}^\Sigma=0.
$$
The same conclusion follows from linearized nondegeneracy when the equality compactification is represented, the nonlinear equality remainder is higher order in the quotient topology, and the linearized equality operator has kernel contained in the propagated closed ledger.

**Proof.** Project the separator identity to the quotient. The closed term disappears, and zero is a sum of nonnegative terms. Positive definiteness forces the open component of the saturation remainder to vanish. For the linearized statement, a nonzero open saturation class has a normalized compactified tangent by the represented equality compactification. The higher-order remainder vanishes in the tangent limit, so this tangent lies in the kernel of the linearized equality operator modulo the closed ledger. Linearized nondegeneracy excludes such a tangent, hence the open saturation quotient is zero. $\square$

### Theorem 8.5 (Hamiltonian Calculation Normal Form)
Every represented Hamiltonian calculation whose purpose is to close a spectral remainder, prove a sign/flux inequality, prove a positive gap, route a degenerate channel, or classify a borderline equality branch has one of the following fast-track normal forms:

1. target-complete quotient vanishing;
2. sharp resistance/gap/passivity radius;
3. commutator-routing rank or positive-commutator certificate;
4. equality-rigidity quotient or separator.

## 10b. Conditional Expectations, Observable Algebras, and Expectation Gauges

This section imports the expectation-gauge material in a form compatible with the Hamiltonian notebook. The ambient observable algebra may be noncommutative; the affordable target readouts used by the framework are represented abelian subalgebras or represented scalar functionals. Nothing outside those represented readouts counts as target progress unless it is supplied as a paid witness.

### Definition 10b.1 (Hamiltonian Observable-Algebra Row)
A Hamiltonian observable-algebra row consists of a represented $\sigma$-finite von Neumann algebra $\mathcal M$ with faithful normal state or trace $\varphi$, a represented budget-generated subalgebra $\mathcal M_{\mathcal B}\subset\mathcal M$, an abelian target-readout subalgebra $\mathcal C_\Sigma\subset\mathcal M_{\mathcal B}$, and a gauge/readout subalgebra $\mathcal N$ containing the declared gauge observables and $\mathcal C_\Sigma$. A conditional-expectation certificate is a normal unital completely positive idempotent
$$
E_{\mathcal N}:\mathcal M\to\mathcal N
$$
which is $\varphi$-preserving and $\mathcal N$-bimodular. In finite-dimensional or tracial cases this is the ordinary trace-preserving projection onto $\mathcal N$ when it exists. In general its existence is part of the row certificate, or follows from the standard modular-invariance criterion for the subalgebra. The GNS space is denoted $L^2(\mathcal M,\varphi)$, with inner product $\langle a,b\rangle_\varphi=\varphi(a^*b)$.

### Lemma 10b.2 (Conditional Expectation is the Hamiltonian Gauge Projection)
The map $E_{\mathcal N}$ extends by continuity to a contraction $P_{\mathcal N}$ on $L^2(\mathcal M,\varphi)$. Its range is $L^2(\mathcal N,\varphi)$, it is idempotent, and
$$
L^2(\mathcal M,\varphi)=L^2(\mathcal N,\varphi)\oplus L^2(\mathcal N,\varphi)^\perp,
\qquad
a=P_{\mathcal N}a+(I-P_{\mathcal N})a.
$$
For every represented readout $c\in\mathcal N$,
$$
\langle c,(I-P_{\mathcal N})a\rangle_\varphi=0.
$$
Thus $P_{\mathcal N}$ is the gauge/readout projection seen by the Hamiltonian target algebra.

**Proof.** Kadison's inequality for the unital completely positive map gives
$$
E_{\mathcal N}(a)^*E_{\mathcal N}(a)\le E_{\mathcal N}(a^*a).
$$
Applying the $\varphi$-preserving state yields $\|E_{\mathcal N}a\|_{2,\varphi}\le\|a\|_{2,\varphi}$, so $E_{\mathcal N}$ extends to a contraction on the GNS completion. Idempotence gives range $L^2(\mathcal N,\varphi)$. If $c\in\mathcal N$, bimodularity and $\varphi$-preservation give
$$
\langle c,a-E_{\mathcal N}a\rangle_\varphi
=\varphi(c^*a)-\varphi(c^*E_{\mathcal N}a)
=\varphi(E_{\mathcal N}(c^*a))-\varphi(c^*E_{\mathcal N}a)=0.
$$
Hence the kernel is orthogonal to the represented readout subspace, and the displayed decomposition is the Hilbert-space projection decomposition after closure. $\square$

### Theorem 10b.3 (Noncommutative/Abelian Exhaustiveness by Expectation)
Assume the Hamiltonian signature supplies the observable-algebra row and that target progress is read only through $\mathcal C_\Sigma\subset\mathcal N$ unless a paid witness is added. Then every observable $a\in L^2(\mathcal M,\varphi)$ has exactly two components for budgeted target progress:
$$
a=P_{\mathcal N}a+(I-P_{\mathcal N})a.
$$
The first component is represented gauge/readout structure. The second component lies in $J_{\rm res}^{0\Sigma}$ for all target readouts in $\mathcal C_\Sigma$. If some functional outside $\mathcal N$ detects $(I-P_{\mathcal N})a$ and is declared target-relevant, that functional is an additional affordable target witness and its representation cost is charged to $\mathcal B$; the component is then no longer residual for the enlarged interface.

**Proof.** The decomposition is Lemma 10b.2. For $c\in\mathcal C_\Sigma\subset\mathcal N$, orthogonality gives $\langle c,(I-P_{\mathcal N})a\rangle_\varphi=0$, which is exactly zero affordable target readout. By affordable target-completeness, zero-readout components are ledgered as $J_{\rm res}^{0\Sigma}$. A nonzero pairing can only be used by adding the pairing functional to the represented target interface. The budget ledger then charges that witness and moves the component from residual to represented structure. These alternatives are exhaustive because $P_{\mathcal N}+I-P_{\mathcal N}=I$ on the GNS space. $\square$

### Definition 10b.4 (Exact and Expectation Gauge in a Hamiltonian Row)
Let $T_\omega$ be the represented quenched Hamiltonian/generator action in a stochastic or random-environment row, and let
$$
\bar T=\mathbb E_\omega[T_\omega]
$$
be its annealed action on the same reduced quotient. A represented direction $\xi$ is an exact gauge if
$$
T_\omega\xi=0\quad\text{for }\gamma\text{-a.e. }\omega,
\qquad
\xi\in\ker\mathfrak q^{\mathsf s}_\omega,
\qquad
\ell_\Sigma(\xi)=0
$$
for every affordable target readout $\ell_\Sigma$. It is an expectation gauge if
$$
\bar T\xi=0,
\qquad
\delta_\omega\xi:=T_\omega\xi-\bar T\xi\in L^2(\gamma;\mathcal H_{\rm red}),
\qquad
\mathbb E_\omega[\delta_\omega\xi]=0,
$$
and $\delta_\omega\xi$ is not identically zero. Exact gauges are quotiented pathwise. Expectation gauges are quotiented for the annealed action only after their quenched fluctuation row is certified.

### Theorem 10b.5 (Annealed Closure and Quenched Fluctuation Resistance)
Let $\xi$ be an expectation gauge. Assume $\xi\in\ker\bar{\mathfrak q}^{\mathsf s}$, $\xi$ is invariant under the annealed gauge/readout algebra, and $\ell_\Sigma(\xi)=0$ for every affordable target readout. Then $\xi$ is an exact gauge of the annealed Hamiltonian/generator row and belongs to the annealed closed ledger. Moreover,
$$
\mathbb E_\omega\,\ell_\Sigma(\delta_\omega\xi)=0
$$
for every continuous affordable target-flux functional $\ell_\Sigma$, so the fluctuation current is annealed $J_{\rm res}^{0\Sigma}$.

For quenched verdicts, let $b_\omega^\xi=P_{\Sigma,\mathcal B}P_\xi^\perp\delta_\omega\xi$ be the target-active fluctuation after projecting off the gauge direction and all already closed ledger classes. Let $G_\xi$ be the represented transverse restoring form on this quotient. The fluctuation-resistance transcript is
$$
\mathcal R^{\rm fl}_\xi
=\mathbb E_\omega\langle b_\omega^\xi,G_\xi^+ b_\omega^\xi\rangle.
$$
If $G_\xi\ge g_\xi I$ on the reduced quotient and $v_\xi=\mathbb E_\omega\|b_\omega^\xi\|^2$, then
$$
\mathcal R^{\rm fl}_\xi\le v_\xi/g_\xi.
$$
A budget bound on $\mathcal R^{\rm fl}_\xi$, or an equivalent moment-capacity transcript
$$
p_\star^{\rm fl}=\sup\{p\ge1:{\rm Cap}_{\mathcal B}^{(p),\xi}(\Sigma)=0\},
$$
closes precisely the same quenched row as a direct pathwise fluctuation estimate. If the propagated leak has zero target flux or zero budget target capacity, it is closed. If a positive-capacity target-visible fluctuation quotient remains, that quotient is handed to the existing capacity/nonpolarity fast-track through the right-process nonpolarity witness or starting-law lower-bound transcript; it is not a separate Hamiltonian gauge obligation.

**Proof.** The first statement is Definition 10b.4 applied to $\bar T$, together with the kernel and invariance hypotheses. Linearity and integrability give
$$
\mathbb E_\omega\ell_\Sigma(\delta_\omega\xi)=\ell_\Sigma(\mathbb E_\omega\delta_\omega\xi)=0,
$$
so the annealed fluctuation has zero affordable target flux and is residual by target-completeness. For the quenched row, the direct pathwise estimate acts only through the target-active component $b_\omega^\xi$ on the quotient where gauge and closed classes have already been removed. The represented inverse-form moment is exactly the resistance of that drive through the transverse restoring form. The spectral bound $G_\xi\ge g_\xi I$ implies $G_\xi^+\le g_\xi^{-1}I$ on the range, hence $\mathcal R^{\rm fl}_\xi\le v_\xi/g_\xi$. Capacity monotonicity gives the down-set defining $p_\star^{\rm fl}$. Zero target flux or zero capacity places the leak in the closed ledger; a positive-capacity target-visible remainder is not residual and is exactly the open quotient. Thus the fluctuation-resistance or moment-capacity transcript is equivalent to the explicit quenched fluctuation calculation on the same quotient. $\square$

**Proof.** The variance formula is the standard solution of the one-dimensional Lyapunov equation $2g_\xi\operatorname{Var}(Y)=\sigma_\xi^2$.
